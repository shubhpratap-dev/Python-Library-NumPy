# 1. Shuffle():

1D Array:

```python
import numpy as np

var = np.array([1,2,3,4,5])

np.random.shuffle(var)

print(var)
```

2D Array:

```python
var = np.array([[1,2],[3,4],[5,6]])

np.random.shuffle(var)

print(var)
```

# 2. Unique():

```python
var1 = np.array([1,2,3,4,2,5,2,4,1,6,7])

print(np.unique(var1))
print(np.unique(var1,return_index=True))
print(np.unique(var1,return_index=True,return_counts=True))
```

# 3. Resize():

```python
var2 = np.array([1,2,3,4,5,6])

y = np.resize(var2,(2,3))   # np.resize(array,(no. of row, no. of col))

print(y)
```

# 4. Flatten():

```python
var2 = np.array([1,2,3,4,5,6])

y = np.resize(var2,(2,3))
print(y)
print()
print(y.flatten())  # By Default order is: 'C'
print()
print("Flatten: ",y.flatten(order="F"))   
print()
print("Ravel: ",y.ravel())
print()
print("Ravel: ",np.ravel(y,order="F"))
```
