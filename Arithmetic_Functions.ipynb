# Arithmetic Operations:

1. Min(), Max() & their Positions (1-D Array):

```python
import numpy as np

var = np.array([3,1,5,8,2,14,6])

print("Min: ",np.min(var), "Position: ", np.argmin(var))
print("Max: ",np.max(var), "Position: ", np.argmax(var))
```

2. Min(), Max() & their Positions (2-D Array):

```python
var1 = np.array([[3,1,5,8,2,14,6],[2,0,4,6,8,1,20]])

print("Min taken as complete: ",np.min(var1))
print("Max taken as complete: ",np.max(var1))
print()
print("Min & Max in each column:")
print("Min: ",np.min(var1,axis=0))  # It gives min in each row
print("Max: ",np.max(var1,axis=0))  # It gives max in each row
print()
print("Min & Max in each row:")
print("Min: ",np.min(var1,axis=1))  # It gives min in each row
print("Max: ",np.max(var1,axis=1))  # It gives max in each row
```

2. SQRT()

```python
var = np.array([3,1,5,8,2,14,6])

print(np.sqrt(var))
```

3. Sin() & Cos()

```python
var2 = np.array([3,1,5,8,2,14,6])

print(np.sin(var2))
print(np.cos(var2))
```

4. Cumunity Sun()

```python
var = np.array([3,1,5,8,2,14,6])

print(np.cumsum(var2))
```
