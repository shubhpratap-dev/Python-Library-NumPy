# 1. Search():

```python
import numpy as np

var = np.array([4,2,3,4,2,5,2,5,6,7])

x = np.where(var == 2)
print(x)
print()

x = np.where((var%2) == 0)
print(x)
print()
```

# 2. Search-Sorted():

```python
var1 = np.array([1,2,3,4,6,7,8,9,10])

x1 = np.searchsorted(var1,5)  # By Default it move left to right
print(x1)
print()

x2 = np.searchsorted(var1,5,side="right")  #  Now it move right ot left
print(x2)
print()

var2 = np.array([1,2,3,8,9,10])
x1 = np.searchsorted(var2,[4,5,6,7],side='right')  # By Default it move left to right
print(x1)
print()
```
