How to create array using list in numpy:

```python
import numpy as np
x = np.array([1,2,3,4,5])  # Here we are passing list in array function
print(x)
print(type(x))
```

Printing list:

```python
y = [1,2,3,4,5]
print(y)
print(type(y))
```

To calculate the run time of a list which was created using for loop:

```python
%timeit z = [j**4 for j in range(1,9)]
print(z)
```

To calculate the run time for of numpy array:

```python
import numpy as np
%timeit a = np.arange(1,9)**4  # here this arange() is used to create array same as in range() way
```

So, from above two codes it proves that numpy is fater than list
