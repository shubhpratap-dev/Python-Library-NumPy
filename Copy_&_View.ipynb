# Copy & View:

```python
import numpy as np

var = np.array([1,2,3,4])

co = var.copy()

print("var: ",var)
print("copy: ",co)
```

```python
x = np.array([9,8,7,6,5])

vi = x.view()

print("x: ",x)
print("view: ", vi)
```

Now on changing data:

```python
var = np.array([1,2,3,4])

co = var.copy()

var[1] = 40

print("var: ",var)
print("copy: ",co)
```

```python
x = np.array([9,8,7,6,5])

vi = x.view()

x[1] = 40

print("x: ",x)
print("view: ", vi)
```
