# Random Array:

# 1. Rand()

```python
import numpy as np

var = np.random.rand(5)
print(var)
```

2-D Array:

```python
var1 = np.random.rand(2,5)   # Array of size(2,5)
print(var1)
```

# 2. Randn():

```python
var2 = np.random.randn(5)
print(var2)
```

# 3. Ranf():

```python
var3 = np.random.ranf(4)
print(var3)
```

# 4. Randint():

```python
var4 = np.random.randint(5,20,5)     # np.random.randint(from,to,no. of elements want to generate)
print(var4)
```
