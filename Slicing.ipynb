# Slicing:

1D Array:

```python
import numpy as np

var = np.array([9,8,7,6,5,4,3,2])
# Index:        0,1,2,3,4,5,6

print(var)
print()

print("2 to 5: ",var[1:6])
print("2 to 5: ",var[1:6:1])
print(var[1:6:2])
print(var[1:])
print(var[:5])
print(var[::2])
print(var[::-1])
print(var[::])
```

2D Array:

```python
var1 = np.array([[1,2,3,4,5],[9,8,7,6,5],[11,12,13,14,15]])

print(var1)
print()
print(var1[1,1:])
print(var1[2,1:])
```
