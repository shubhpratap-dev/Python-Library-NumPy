# Spilt Array:

Splitting 1D Array:

```python
import numpy as np

var = np.array([1,2,3,4,5,6])
print(var)
print()

ar = np.array_split(var,2)
print(ar)
print(type(ar))
print()

arr = np.array_split(var,3)
print(arr)
print(type(arr))
print(arr[0],type(arr[0]))
```

Splitting 2D Array:

```python
var1 = np.array([[1,2],[3,4],[5,6]])
print(var1)
print()

ar = np.array_split(var1,2)
print(ar)
print(type(ar))
print()

arr = np.array_split(var1,3)
print(arr)
print(type(arr))
print(arr[0],type(arr[0]))
print()

ar2 = np.array_split(var1,3,axis=1)
print(ar2)
```
