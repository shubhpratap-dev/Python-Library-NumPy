# Join Array:

Joining 1D Array: Using Concatenate()

```python
import numpy as np

var = np.array([1,2,3,4])
var1 = np.array([9,8,7,6])

ar = np.concatenate((var,var1))
print(ar)
```

```python
var = np.array([1,2,3,4,5])
var1 = np.array([9,8,7,6])

ar = np.concatenate((var,var1))
print(ar)
```

Joining 2D Array: Using Concatenate()

```python
var2 = np.array([[1,2],[3,4]])
var3 = np.array([[9,8],[7,6]])
print(var2)
print()
print(var3)
print()

arr = np.concatenate((var2,var3),axis=0)
print(arr)
print()

arrr = np.concatenate((var2,var3),axis=1)
print(arrr)
```

Joining 1D Array: Using Stack()

```python
var4 = np.array([1,2,3,4])
var5 = np.array([9,8,7,6])

ar_new = np.stack((var4,var5))
print(ar_new)
print()

ar_new = np.stack((var4,var5),axis=0)
print(ar_new)
print()

ar_new = np.stack((var4,var5),axis=1)
print(ar_new)
print()

ar_new = np.hstack((var4,var5))   # Work along Row
print(ar_new)
print()

ar_new = np.vstack((var4,var5))   # Work along Column
print(ar_new)
print()

ar_new = np.dstack((var4,var5))   # Work along Height
print(ar_new)
print()
```
