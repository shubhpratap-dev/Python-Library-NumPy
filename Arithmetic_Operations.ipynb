# Aritmetic Operations:

# (1.) 1-D Array:

Adding elements of array with an another elemnt:

```python
import numpy as np

var = np.array([1,2,3,4])

varadd = var + 3
print(varadd)
```

```python
var = np.array([1,2,3,4])

varadd = np.add(var,3)
print(varadd)
```

Adding two different array:

```python
var1 = np.array([1,2,3,4])
var2 = np.array([1,2,3,4])
varadd = var1 + var2
print(varadd)
```

```python
var1 = np.array([1,2,3,4])
var2 = np.array([1,2,3,4])
varadd = np.add(var1,var2)
print(varadd)
```

Subtraction:

```python
var = np.array([1,2,3,4])

varadd = var - 3
print(varadd)
```

Multiplication

```python
var = np.array([1,2,3,4])

varadd = var * 3
print(varadd)
```

Divide:

```python
var = np.array([1,2,3,4])

varadd = var/3
print(varadd)
```

Modulus:

```python
var = np.array([1,2,3,4])

varadd = var%3
print(varadd)
```

Reciprocal:

```python
var = np.array([1,2,3,4])

varadd = np.reciprocal(var)
print(varadd)
```

# (2) 2-D Array:

Add:

```python
var21 = np.array([[1,2,3,4],[1,2,3,4]])
var22 = np.array([[1,2,3,4],[1,2,3,4]])

print(var21)
print()
print(var22)
print()

varadd2 = var21 + var22

print(varadd2)
```

Multiply:

```python
var21 = np.array([[1,2,3,4],[1,2,3,4]])
var22 = np.array([[1,2,3,4],[1,2,3,4]])

print(var21)
print()
print(var22)
print()

varadd2 = var21 * var22

print(varadd2)
```
