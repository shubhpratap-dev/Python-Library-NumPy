# Zero Array:

```python
import numpy as np
ar_zero = np.zeros(4)
ar_zero1 = np.zeros((3,4))    # Array of size (3,4) with 0 elements created
print(ar_zero)
print()
print(ar_zero1)
```

# One's Array:

```python
ar_one = np.ones(4)
print(ar_one)
```

# Empty Array:

```python
ar_em = np.empty(4)
print(ar_em)         # But here it show array of 1 bcz in empty array the elements of previous array stored
```

# Range Array:

```python
ar_rn = np.arange(4)    # we can also give value as: np.arange(0,4)
print(ar_rn)
```

# Digonal Array:

```python
ar_dia = np.eye(3)
print(ar_dia)
```

```python
ar_dia = np.eye(3,5)
print(ar_dia)
```

# Linespace Array:

```python
ar_lin = np.linspace(0,20,num=5)   # np.linspace(from,to,no.of elements)
print(ar_lin)
```
