```python
import numpy as np

var1 = np.array([1,2,3,4])
var2 = np.array([1,2,3])
print(var1 + var2)
```

This is broadcast error

```python
var1 = np.array([1,2,3])
print(var1)
print(var1.shape)
print()

var2 = np.array([[1],[2],[3]])
print(var2)
print(var2.shape)
print()

print(var1 + var2)
```
